```python
### Main Widget Usage ###

# Usage order: Lasso Select --> Confirm Selection --> Extract Spectra (--> Reset)
try:
    qw.close()
except:
    pass

qw = QueryWidget(coords = SkyCoord(32.52204* u.deg, -0.04762* u.deg, frame="icrs"),
                 #detectid = 3010418916, # an LAE in Laurel's PLAGUE sample
                 #detections = detects, #need to turn this on if using detectid!!
                 cutout_size=10*u.arcsec, 
                 spec_mode='fibers', # 'fibers', 'psf'
                 parallel_shotids = [20250821018, 20250825020, 20250828010, 20250829018, 20250921021, 20250925009, 20250823020, 20250926017, 20251015012, 20250822014],
               #  wave_range = (4000, 5000),
               #  flux_range = (-1, 1),
                 stat='weighted_biweight', # 'mean', 'median', 'biweight', 'weighted_biweight'
                 )

# Image loading is not always immediate. matplotlib has to redraw the figure for every pan/lasso draw. please be patient. 

# To select different fibers, simply redraw the lasso or hit reset. However, the reset button will pan back to original coords. 
# Reset takes a second to load. 

# NOTE: if you use the up/down arrows on RA/Dec, it will truncate to the set step (i.e. be less exact). 
# The exact selected fiber coordinates are saved in the output table.
```
